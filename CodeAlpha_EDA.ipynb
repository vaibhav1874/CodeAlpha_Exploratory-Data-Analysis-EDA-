### 18. Summary Statistics
Let's calculate summary statistics using `df.describe()` for numerical features. We also compute and print standard deviation, skewness, and kurtosis to understand distributions.

```python
# Standard describe summary
display(df.describe())
```

```python
# Print detailed statistics for each numerical column
numeric_columns = ['Experience_Years', 'Daily_Token_Usage', 'Tasks_Automated_Per_Week', 'Productivity_Gain_Percent']

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    
    print(f"=== Statistics for '{col}' ===")
    print(f"  Mean: {df[col].mean():.2f}")
    print(f"  Median: {df[col].median():.2f}")
    print(f"  Mode: {df[col].mode().iloc[0]:.2f}")
    print(f"  Minimum: {df[col].min():.2f}")
    print(f"  Maximum: {df[col].max():.2f}")
    print(f"  Std Dev: {df[col].std():.2f}")
    print(f"  IQR: {iqr:.2f} (Q1: {q1:.2f}, Q3: {q3:.2f})")
    print(f"  Skewness: {df[col].skew():.2f}")
    print(f"  Kurtosis: {df[col].kurt():.2f}")
    print("-" * 40)
```

**Observation**:
- **Experience_Years**: Mean (13.01) and median (13.0) are equal, indicating a symmetric, flat distribution (skewness 0.00).
- **Daily_Token_Usage**: Average token usage is ~4,775 per day, ranging from 500 to 10,000.
- **Tasks_Automated_Per_Week**: Average is exactly 10 tasks per week. Distribution is normal-like.
- **Productivity_Gain_Percent**: Average productivity gain is 11.22%. However, the maximum is 84.90% and median is 8.90%. Highly skewed right (skewness 1.81) and heavy-tailed (kurtosis 3.78), implying the presence of high-impact outliers.

### 19. Explore Numerical Features
We plot the distribution (Histogram) and spread (Boxplot) of our four numerical columns to understand their shapes.

```python
# Experience_Years distribution
plt.figure(figsize=(12, 4.5))
plt.subplot(1, 2, 1)
sns.histplot(df['Experience_Years'], kde=True, color='royalblue')
plt.title('Distribution of Experience Years')
plt.xlabel('Experience Years')
plt.ylabel('Count')

plt.subplot(1, 2, 2)
sns.boxplot(x=df['Experience_Years'], color='lightskyblue')
plt.title('Boxplot of Experience Years')
plt.xlabel('Experience Years')

plt.tight_layout()
plt.show()
```

**Observation**: Experience years are evenly distributed across the 1 to 25 year range. The boxplot confirms that the median is 13 years, and there are no outliers.

```python
# Daily_Token_Usage distribution
plt.figure(figsize=(12, 4.5))
plt.subplot(1, 2, 1)
sns.histplot(df['Daily_Token_Usage'], kde=True, color='seagreen')
plt.title('Distribution of Daily Token Usage')
plt.xlabel('Daily Token Usage')
plt.ylabel('Count')

plt.subplot(1, 2, 2)
sns.boxplot(x=df['Daily_Token_Usage'], color='lightgreen')
plt.title('Boxplot of Daily Token Usage')
plt.xlabel('Daily Token Usage')

plt.tight_layout()
plt.show()
```

**Observation**: Daily token usage spans evenly from 500 to 10,000 tokens, showing standard usage levels with no extreme values.

```python
# Tasks_Automated_Per_Week distribution
plt.figure(figsize=(12, 4.5))
plt.subplot(1, 2, 1)
sns.histplot(df['Tasks_Automated_Per_Week'], kde=True, color='crimson')
plt.title('Distribution of Weekly Automated Tasks')
plt.xlabel('Automated Tasks Per Week')
plt.ylabel('Count')

plt.subplot(1, 2, 2)
sns.boxplot(x=df['Tasks_Automated_Per_Week'], color='salmon')
plt.title('Boxplot of Weekly Automated Tasks')
plt.xlabel('Automated Tasks Per Week')

plt.tight_layout()
plt.show()
```

**Observation**: Weekly automated tasks have a symmetric, normal-like bell curve peaking at 10. No outliers are observed.

```python
# Productivity_Gain_Percent distribution
plt.figure(figsize=(12, 4.5))
plt.subplot(1, 2, 1)
sns.histplot(df['Productivity_Gain_Percent'], kde=True, color='purple')
plt.title('Distribution of Productivity Gain Percent')
plt.xlabel('Productivity Gain (%)')
plt.ylabel('Count')

plt.subplot(1, 2, 2)
sns.boxplot(x=df['Productivity_Gain_Percent'], color='violet')
plt.title('Boxplot of Productivity Gain Percent')
plt.xlabel('Productivity Gain (%)')

plt.tight_layout()
plt.show()
```

### 20. Explore Categorical Features
Let's visualize counts across categorical features: Industry, Location, Primary_AI_Tool, and Job_Role.

```python
# Industry and Location count plots
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
sns.countplot(data=df, y='Industry', order=df['Industry'].value_counts().index, palette='viridis', hue='Industry', legend=False)
plt.title('User Counts by Industry')
plt.xlabel('Number of Users')

plt.subplot(1, 2, 2)
location_counts = df['Location'].value_counts()
plt.pie(location_counts, labels=location_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Location Share %')

plt.tight_layout()
plt.show()
```

**Observation**: The user distribution across the 6 industries and 5 locations is balanced. No single industry or region dominates the sample.

```python
# Primary AI Tool count plot
plt.figure(figsize=(10, 5))
sns.countplot(data=df, y='Primary_AI_Tool', order=df['Primary_AI_Tool'].value_counts().index, palette='flare', hue='Primary_AI_Tool', legend=False)
plt.title('Usage of Primary AI Tools')
plt.xlabel('Number of Users')
plt.ylabel('AI Tool')
plt.tight_layout()
plt.show()
```

```python
# Job Role count plot
plt.figure(figsize=(10, 8))
sns.countplot(data=df, y='Job_Role', order=df['Job_Role'].value_counts().index, palette='magma', hue='Job_Role', legend=False)
plt.title('User Counts by Job Role')
plt.xlabel('Number of Users')
plt.ylabel('Job Role')
plt.tight_layout()
plt.show()
```
