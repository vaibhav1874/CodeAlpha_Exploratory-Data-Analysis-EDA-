### 18. Summary Statistics
I will use `df.describe()` to see summary statistics for numerical features, and print detailed calculations for each numerical column.

```python
# Print detailed statistics for each numerical column
numeric_columns = ['Experience_Years', 'Daily_Token_Usage', 'Tasks_Automated_Per_Week', 'Productivity_Gain_Percent']

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    
    print(f"=== Statistics for '{col}' ===")
    print(f"  Mean: {df[col].mean():.2f}")
    print(f"  Median: {df[col].median():.2f}")
    print(f"  Mode: {df[col].mode().iloc[0]:.2f}")
    print(f"  Minimum: {df[col].min():.2f}")
    print(f"  Maximum: {df[col].max():.2f}")
    print(f"  Std Dev: {df[col].std():.2f}")
    print(f"  IQR: {iqr:.2f} (Q1: {q1:.2f}, Q3: {q3:.2f})")
    print(f"  Skewness: {df[col].skew():.2f}")
    print(f"  Kurtosis: {df[col].kurt():.2f}")
    print("-" * 40)
```

### 24. Outlier Detection
We calculate productivity outliers using the Interquartile Range (IQR) method.

```python
# Find outliers using IQR cutoff
q1 = df['Productivity_Gain_Percent'].quantile(0.25)
q3 = df['Productivity_Gain_Percent'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[(df['Productivity_Gain_Percent'] < lower_bound) | (df['Productivity_Gain_Percent'] > upper_bound)]
print(f"Productivity Gain Upper Outlier Cutoff: {upper_bound:.2f}%")
print(f"Number of outliers: {len(outliers)}")
```

### Decision: KEEP Outliers

**Why:**
1. **Real Data**: These data points represent genuine "power users" who automate multiple repetitive workflows and achieve massive productivity gains (up to 84.9%). They are not errors.
2. **Valuable Insights**: Removing them would hide the most interesting and successful cases of AI tool adoption.

### 26. Test Simple Hypotheses
**Hypothesis 1:** "Users who automate more tasks per week achieve higher productivity gains."

```python
# Mean productivity gain grouped by weekly automated tasks
tasks_group = df.groupby('Tasks_Automated_Per_Week')['Productivity_Gain_Percent'].mean().reset_index()
plt.figure(figsize=(8, 5))
sns.lineplot(data=tasks_group, x='Tasks_Automated_Per_Week', y='Productivity_Gain_Percent', marker='o', color='blue')
plt.title('Weekly Automated Tasks vs Mean Productivity Gain')
plt.xlabel('Tasks Automated')
plt.ylabel('Mean Productivity Gain (%)')
plt.tight_layout()
plt.show()
```

**Verdict: Accept Hypothesis 1.** The line chart shows a strong upward trend, and the correlation is high (0.70).

**Hypothesis 2:** "More experienced users consume more daily tokens."

```python
# Mean daily token usage grouped by experience years
exp_group = df.groupby('Experience_Years')['Daily_Token_Usage'].mean().reset_index()
plt.figure(figsize=(8, 5))
sns.scatterplot(data=exp_group, x='Experience_Years', y='Daily_Token_Usage', color='orange', s=100)
plt.title('Experience Years vs Mean Daily Token Usage')
plt.xlabel('Experience Years')
plt.ylabel('Mean Daily Token Usage')
plt.tight_layout()
plt.show()
```

**Verdict: Reject Hypothesis 2.** The scatter plot is flat, showing that average token usage stays around 4,800 tokens regardless of experience level. The correlation is near 0.0.
